```python
import colorcet
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import plotly.express as px
import plotly.graph_objects as go
import plotly.io as pio
import plotly.colors as colors

```

```python
df=pd.read_csv("D:/data science/data analysis/E-Commerce/Sample - Superstore.csv",encoding="latin1")
```

```python
df.head()
```

```python
df.info()
```

convert order date and ship date format from object to datetime

```python
df['Order Date']=pd.to_datetime(df['Order Date'])
df['Ship Date']=pd.to_datetime(df['Ship Date'])
df.head()
```

**Now we are going to add 3 new coloum for better analysis (order month ow.oy)**


```python
df['order_month']=df['Order Date'].dt.month
df['order_year']=df['Order Date'].dt.year
df['order day of week']=df['Order Date'].dt.dayofweek
df.head()
```

**QUESTION** ::   *YOU NEED TO CALCULATE THE MONTHLY SALE OF THE STORE AND IDENTIFY WHICH MONTH HAD THE HIGHEST SALE AND WHICH MONTH HAD LOWEST SALE*

```python
sales_by_month=df.groupby('order_month')['Sales'].sum().reset_index()
sales_by_month
```

```python
#plt.plot(sales_by_month)
#plt.show()
px.line(sales_by_month,x='order_month',y='Sales',title='Sales by Month',markers='0')

```

**Question2**
You need to analyze sales based on product categories and determine which category has
the lowest sales and which category has the highest sales.


```python
sales_by_category=df.groupby('Category')['Sales'].sum().reset_index()
sales_by_category
```

```python
fig=px.pie(sales_by_category,values='Sales',names='Category',title='Categories',hole=0.5, color_discrete_sequence=px.colors.qualitative.Pastel1)
fig.update_traces(textposition='inside',textinfo='percent+label')
fig.update_layout(title_text='Sales by Category')

fig.show()
```

**QUESTION3**The sales analysis needs to be done based on sub-categories

```python
sales_by_subcategory=df.groupby(['Sub-Category'])['Sales'].sum().reset_index()
sales_by_subcategory
```

```python
fig1=px.bar(sales_by_subcategory,x='Sub-Category',y='Sales',title='Sales by Sub-Category',color='Sub-Category')
fig1.update_traces(textposition='inside')
```

```python
fig1.show()
```

**QUESTION:4** You need to analyze the monthly profit from sales and determine which month had the

```python
profit_by_month=df.groupby('order_month')['Profit'].sum().reset_index()
profit_by_month
```

**QUESTION : 5** Analyze the sales and profit by customer segmen


```python
profit_By_segment = df.groupby(['Segment']).agg({'Sales': 'sum', 'Profit': 'sum'}).reset_index()

color_palette = px.colors.qualitative.Pastel
fig5 = go.Figure()
fig5.add_trace(go.Bar(
    x=profit_By_segment['Segment'],
    y=profit_By_segment['Sales'],
    marker_color=color_palette[0]
))
fig5.add_trace(go.Bar(x=profit_By_segment['Segment'], y=profit_By_segment['Profit'], marker_color=color_palette[1]))
fig5.show()
```

**QUESTION:6** Analyze the sales to profit ratio

```python
sales_profit_by_segment = df.groupby(['Segment']).agg({'Sales': 'sum', 'Profit': 'sum'}).reset_index()
sales_profit_by_segment['sales_to_ratio']=sales_profit_by_segment['Sales'] / sales_profit_by_segment['Profit']
print(sales_profit_by_segment[['Segment','sales_to_ratio']])


```
